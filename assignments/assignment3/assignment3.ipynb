## Question 3 - **(8 marks)** Floating Point Error Analysis

Show that if $x$ is a representable number in some floating point system $F$, the relative error
of the floating point calculation $(x\otimes x)\ominus 1$ compared to the true value of $x^2-1$ is less than
or equal to

$$\left(\frac{|x^2|}{|x^2-1|}+1\right)E \tag{2}$$

assuming we can drop/ignore terms proportional to $E^2$ and higher powers of $E$, and no overflow/underflow occurs. (Recall that the circle notation indicates a floating point counterpart
of a corresponding real arithmetic operator.)



**Answer:**

Since $x$ is representable in $F$, we can write each floating-point operation as its exact counterpart multiplied by $(1+\eta_i)$, where $\lvert\eta_i\rvert\leq E$.

First, the floating-point multiplication gives:

\begin{align*}
x\otimes x = x^2(1+\eta_1)
\end{align*}

Then, applying floating-point subtraction:

\begin{align*}
(x\otimes x)\ominus 1
&= (x^2(1+\eta_1)-1)(1+\eta_2)\\
&= (x^2+x^2\eta_1-1)(1+\eta_2)\\
&= x^2-1+x^2\eta_1+(x^2-1)\eta_2+x^2\eta_1\eta_2
\end{align*}

Subtracting the true value $x^2-1$, we obtain the absolute error:

\begin{align*}
\text{Absolute error}
&= \lvert (x\otimes x)\ominus 1-(x^2-1)\rvert\\
&= \lvert x^2\eta_1+(x^2-1)\eta_2+x^2\eta_1\eta_2\rvert\\
&\leq \lvert x^2\rvert\lvert\eta_1\rvert
+\lvert x^2-1\rvert\lvert\eta_2\rvert
+\lvert x^2\rvert\lvert\eta_1\rvert\lvert\eta_2\rvert\\
&\leq \lvert x^2\rvert E+\lvert x^2-1\rvert E+\lvert x^2\rvert E^2
\end{align*}

As permitted by the question, we ignore terms proportional to $E^2$ and higher powers:

\begin{align*}
\text{Absolute error}
\leq \lvert x^2\rvert E+\lvert x^2-1\rvert E
\end{align*}

Dividing by the absolute value of the true result, assuming $x^2\neq 1$, gives the relative error:

\begin{align*}
\text{Relative error}
&= \frac{\lvert (x\otimes x)\ominus 1-(x^2-1)\rvert}{\lvert x^2-1\rvert}\\
&\leq \frac{\lvert x^2\rvert E+\lvert x^2-1\rvert E}{\lvert x^2-1\rvert}\\
&= \frac{\lvert x^2\rvert}{\lvert x^2-1\rvert}E+E\\
&= \boxed{\left(\frac{\lvert x^2\rvert}{\lvert x^2-1\rvert}+1\right)E}
\end{align*}


## Question 4 - **(6 marks)** Convergence of Fixed Point Iteration

Consider a fixed point iteration:

$$x_{k+1}=g(x_k) \tag{3}$$

$$g(x)=\frac{\sin(x+\pi/2)}{2} \tag{4}$$

Will this iteration converge for any $x\in[0,\pi/2]$? Justify your answer by considering all the
conditions of a convergence theorem.



**Answer:**

We are given the fixed point iteration

$$
x_{k+1}=g(x_k),\qquad
g(x)=\frac{\sin(x+\pi/2)}{2}=\frac{\cos(x)}{2}
$$

on the interval $I=[0,\pi/2]$.

We apply the Contraction Mapping Theorem, which guarantees convergence for any initial value $x_0\in I$ if $g$ is continuous, maps $I$ into itself, and is a contraction on $I$.

Since $\cos(x)$ is continuous on $\mathbb{R}$, $g(x)=\cos(x)/2$ is continuous on $I$.

For all $x\in[0,\pi/2]$, we have $0\leq\cos(x)\leq1$. Therefore,

\begin{align*}
0 &\leq g(x)\leq\frac12\\
g(I)&\subseteq[0,1/2]\subseteq[0,\pi/2]=I
\end{align*}

Thus, $g$ maps $I$ into itself.

We calculate the derivative:

\begin{align*}
g'(x)&=-\frac{\sin(x)}{2}\\
\lvert g'(x)\rvert&=\frac{\lvert\sin(x)\rvert}{2}
\end{align*}

Since $0\leq\sin(x)\leq1$ for all $x\in I$,

\begin{align*}
\max_{x\in I}\lvert g'(x)\rvert
=\frac12<1
\end{align*}

Therefore, $g$ is a contraction on $I$ with contraction constant $L=1/2$.

All conditions of the Contraction Mapping Theorem are satisfied. Hence, $g$ has a unique fixed point $x^*\in I$, and the iteration $x_{k+1}=g(x_k)$ converges to $x^*$ for every initial value $x_0\in[0,\pi/2]$.

Find the root ($x^*=\sqrt{3}$) of the following equation:

$$x^2-3=0,\qquad x\in[0.5,3.5]$$

using your four root finding functions with `tol=1e-10` and `maxiter=100`. For the bisection
and secant methods, use `[a,b]=[0.5,3.5]`. For Newton and fixed point iteration, use `x0=0.5`.
You will need to determine a $g(x)$ function such that the fixed point iteration converges. There

### Answer - root finding functions


````python
def bisection(f, a, b, tol, maxiter):
    xs = []
    for _ in range(maxiter):
        c = a + (b - a) / 2
        xs.append(c)
        if len(xs) > 1 and abs(xs[-1] - xs[-2]) < tol:
            return xs
        if f(a) * f(c) < 0:
            b = c
        else:
            a = c
    return xs
````

````python
def secant(f, a, b, tol, maxiter):
    xs = []
    prev = a
    curr = b
    for _ in range(maxiter):
        if f(curr) == f(prev):
            return xs

        next_x = curr - f(curr) * ((curr - prev) / (f(curr) - f(prev)))

        xs.append(next_x)
        if abs(next_x - curr) < tol:
            return xs
        prev = curr
        curr = next_x
    return xs
````

````python
def newton(f, fprime, x0, tol, maxiter):
    xs = []
    curr = x0
    for _ in range(maxiter):
        if fprime(curr) == 0:
            return xs

        next_x = curr - f(curr) / fprime(curr)
        xs.append(next_x)
        if abs(next_x - curr) < tol:
            return xs
        curr = next_x
    return xs
````

````python
def fixedpoint(g, x0, tol, maxiter):
    xs = []
    curr = x0
    for _ in range(maxiter):
        next_x = g(curr)
        xs.append(next_x)
        if abs(next_x - curr) < tol:
            return xs
        curr = next_x
    return xs
````
