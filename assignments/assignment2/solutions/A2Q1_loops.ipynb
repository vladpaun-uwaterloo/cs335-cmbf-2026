```python
# Compute Black-Scholes option value using a binomial tree
# European case
# uses loops

import numpy as np

S0 = 100 # S0 - current stock price
K = 100 # K - strike
T = 1 # T - expiry time
r = .02 # r - interest rate
sigma = .3 # sigma - volatility

opttype = 1 # opttype - 0 for a call, otherwise a put
countrytype = 1 #countrytype - 0 for European, 1 for American

for outer_count in np.arange(5):

    Nsteps = 500*2**outer_count; # Nsteps - number of timesteps
    delt = T/Nsteps;

    # tree parameters
    u = np.exp(sigma * np.sqrt(delt) )
    d = 1/u;
    a = np.exp( r*delt )
    p = (a - d)/(u - d)
    
    # payoff at t=T
    V= np.zeros(Nsteps+1);

    # W[j] corresponds to node S_{j}^N
    # j=0,...,Nsteps
    for j in np.arange(0,Nsteps+1):
        V[j] = S0*(u**(j))*(d**(Nsteps - j))

    #vector operations - determine payoff values at N
    if opttype == 0: #call
        V =  np.maximum( V - K, 0)
    else: #put
        V = np.maximum( K - V, 0)

    # backward recursion
    for n in np.arange(Nsteps-1, -1, -1):
        for j in np.arange(n+1): 
            if countrytype == 0: #European
                V[j] = np.exp(-r*delt)*( p*V[j+1] + (1-p)*V[j] )
            else: #American
                Vc = np.exp(-r*delt)*( p*V[j+1] + (1-p)*V[j] )
                Snj = S0*u**(j)*d**(n - j)
                
                if opttype == 0: #call
                    payoff = max( Snj - K, 0);
                else:             #put
                    payoff = max( K - Snj, 0);
                
                V[j] = max(Vc, payoff);

    value = V[0]
    print("Tree value:" + str(value))
```
