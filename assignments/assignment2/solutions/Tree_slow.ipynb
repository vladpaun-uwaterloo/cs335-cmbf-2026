```python
# Compute Black-Scholes option value using a binomial tree
# European case
# uses loops - so it's quite slow

import numpy as np

S0 = 100 # S0 - current stock price
K = 100; # K - strike
T = 1; # T - expiry time
r = .02; # r - interest rate
sigma = .3; # sigma - volatility
opttype = 1; # opttype - 0 for a call, otherwise a put
Nsteps = 1000 # Nsteps - number of timesteps 
              #using a low step count because it's so slow!


delt = T/Nsteps;

#
# tree parameters
#
u = np.exp(sigma * np.sqrt(delt) );
d = 1./u;
a = np.exp( r*delt );
p = (a - d)/(u - d);
    
#
# payoff at t=T
#
W= np.zeros(Nsteps+1);

# W[j] corresponds to node S_{j}^N
# j=0,...,Nsteps
for j in np.arange(0,Nsteps+1):
    W[j] = S0*(u**(j))*(d**(Nsteps - j))

#vector operations - determine payoff values at N
if opttype == 0: #call
    W =  np.maximum( W - K, 0)
else: #put
    W = np.maximum( K - W, 0)

# backward recursion
for n in reversed(range(Nsteps)):
    for j in np.arange(n+1):
        W[j] = np.exp(-r*delt)*( p*W[j+1] + (1-p)*W[j] )
        
print("Tree Value: " + str(W[0]))
```
